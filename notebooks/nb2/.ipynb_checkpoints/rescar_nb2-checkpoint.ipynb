# Caderno #2 de análises com dados do ResCar

Esse caderno é a documentação da criação de um **índice padronizado para medir a evolução das escolas ao longo dos anos em cada quesito e no geral**. No caderno anterior, com o intuito de demonstrar a utilidade dos dados coletados no ETL, foram usadas métricas simples e de rigor flexível, como média de notas por quesito ano a ano, na maioria dos casos. Os resultados foram valiosos, mas para aprofundar a análise da percepção do júri sobre os desfiles dos últimos 30 anos, precisamos fazer a pergunta: **como posicionar uma escola em relação às outras em um único ano, e como comparar as posições das escolas ao longo do tempo?** Não é uma pergunta tão simples quanto parece. Como é sabido, as regras de apuração mudaram muito entre 1996 e 2026, de forma que a comparação se torna menos rigorosa tanto mais simplificações são feitas. Outro ponto é que não estamos tentando entender qual escola foi "melhor" *no geral* em cada ano. Se fosse esse o caso, bastaria o resultado total e a classificação oficial das campeãs de cada carnaval, que leva em conta inclusive penalizações. Queremos entender, em vez disso, o *comportamento dos jurados* de cada quesito em relação a cada escola em cada ano, e visualizar esse comportamento ao longo do tempo. Por esse mesmo motivo fez-se a escolha, já no design do ETL, de ignorar o descarte de notas, quando houver. Portanto, esse próximo passo é apenas o desencadeamento lógico dessa primeira decisão, que permitirá uma análise de dados mais granulares e detalhados.

## Um índice universal

A ideia é usar uma variável chamada aqui de *Diferencial da Escola no Ano* para comparar os resultados. Esse valor será alcançado através de outra variável, o *Diferencial da Escola no Quesito* (ou *DESCq*), calculada ano por ano, quesito por quesito, a partir da *subtração da soma das notas de uma escola em um determinado quesito em um dado ano pela média da nota de todas as escolas naquele quesito e naquele ano*. As fórmulas abaixo explicam a ideia:

$$
DESCq_{e,a,q} = n_{e,a,q} - \bar{n}_{a,q}
$$

Onde:

- $DESCq_{e,a,q}$ representa o diferencial da escola $e$, no ano $a$, no quesito $q$;
- $n_{e,a,q}$ é a nota total da escola naquele ano e quesito;
- $\bar{n}_{a,q}$ é a média das notas de todas as escolas naquele ano e quesito.


E o *Diferencial da Escola no Ano* (ou DESCA) é obtido *somando todos os DESCq da escola para cada quesito em um ano e dividindo pelo número total de quesitos*:

$$
DESCA_{e,a} = \frac{1}{Q} \sum_{q=1}^{Q} DESCq_{e,a,q}
$$

Onde:

- $DESCA_{e,a}$ é o diferencial da escola $e$ no ano $a$;
- $q$ é o índice dos quesitos e $Q$ é o número total de quesitos;

Espera-se que esse coeficiente torne menos obtusa a comparação entre os resultados de competições diferentes ao longo dos anos porque ele: **a)** produz um número único que busca capturar o desempenho de uma escola na percepção dos jurados em um ano, que é o que queremos comparar, **b)** ajusta o cálculo de média para fazer mais justiça às escolas, excluindo a necessidade de ignorar quesitos que deixaram de existir (como o Conjunto), e **c)** pode reduzir os ruídos na comparação entre escolas com mais ou menos tempo no Grupo Especial durante esses 30 anos. Essa terceira hipótese ainda deve ser testada na prática.

## Tradução para o Python

Resta agora traduzir esse raciocínio para um (ou alguns) algoritmos Python capaz(es) de realizar esse cálculo no volume considerável de dados que possuímos.

```python
# Função para calcular DESCq

def descq(data, e, a, q):

    nota_escola_quesito = (data[(data['sigla'] == e) & (data['ano'] == a) & (data['quesito'] == q) & (data['quesito'] != 'Pen.')]['nota'].sum())

    totais_quesito = (data[(data['ano'] == a) & (data['quesito'] == q) & (data['quesito'] != 'Pen.')].groupby('sigla')['nota'].sum())

    media_ano_quesito = totais_quesito.mean()

    return nota_escola_quesito - media_ano_quesito
```

```python
# Função para calcular DESCA

def desca(data, e, a):
    
    quesitos = data[(data['sigla'] == e) & (data['ano'] == a) & (data['quesito'] != 'Pen.')]['quesito'].unique()

    if len(quesitos) == 0:
        return None

    soma_quesitos = 0

    for q in quesitos:
        soma_quesitos += descq(data, e, a, q)

    return soma_quesitos / len(quesitos)
```

## Testes de aplicação do cálculo

Os blocos a seguir têm o objetivo de explorar a aplicação de cálculo do DESCq e do DESCA aos dados extraídos no ETL e, posteriormente, polir a análise com visualizações a partir desses cálculos. O arquivo .csv será o mesmo utilizado no primeiro caderno.

```python
# Importação de libs

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Arquivo de resultados .csv

res = pd.read_csv('rescar_1996-2026.csv')
```

```python
# Iteração pelas escolas aplicando a função desca() e guardando em .csv

descas = []

for esc in res['sigla'].unique():
    
    anos_escola = res[res['sigla'] == esc]['ano'].unique()
    
    for a in anos_escola:
            
        desca_value = desca(res, esc, a)

        desca_dict = {
                'sigla': esc,
                'ano': a,
                'desca': desca_value
            }

        descas.append(desca_dict)


pd.DataFrame(descas).to_csv('desca.csv', index=False)
```

```python
# Arquivo .csv com todos os DESCAs

fdesca = pd.read_csv('desca.csv')
```

### Decompondo o DESCA

Sabemos que o DESCA é na verdade uma média calculada a partir dos DESCq da escola em cada quesito. A próxima pergunta que podemos fazer, então, é: **quanto cada quesito contribuiu para o DESCA?** Isso é o mesmo que perguntar **qual é a responsabilidade de cada quesito por um DESCA positivo ou negativo**. Esse cálculo permitiria entender, por exemplo, se a Acadêmicos de Niterói em 2026, que recebeu um DESCA menor que -0.6, teve problemas que se espalharam por todos os quesitos na percepção dos jurados ou se foram problemas mais focalizados. A forma mais simples de fazer isso é simplesmente calcular a porcentagem que cada DESCq representa na soma de todos eles para uma escola em um ano específico.

```python
# Função que retorna os DESCq discriminados para uma escola em um dado ano

def disc_descq(data, e, a):
    
    data_filtrado = data[
        (data['sigla'] == e) &
        (data['ano'] == a) &
        (data['quesito'] != 'Pen.')]

    if data_filtrado.empty:
        return None

    quesitos = data_filtrado['quesito'].unique()
    
    descq_dict = {
        q: descq(data, e, a, q)
        for q in quesitos
    }
    
    return descq_dict
```

```python
# DFs com DESCq discriminado por escola/ano

disc_descq_nit = pd.DataFrame(disc_descq(res, 'NIT', 2026).items(), columns=['quesito', 'descq'])
disc_descq_tui = pd.DataFrame(disc_descq(res, 'TUI', 2017).items(), columns=['quesito', 'descq'])
disc_descq_vir = pd.DataFrame(disc_descq(res, 'VIR', 2026).items(), columns=['quesito', 'descq'])
disc_descq_por = pd.DataFrame(disc_descq(res, 'POR', 2017).items(), columns=['quesito', 'descq'])
```

```python
# Gráfico 4: Niterói/2026: DESCq discriminado

plt.figure(figsize=(10,6))

sns.barplot(
    data=disc_descq_nit,
    x='descq',
    y='quesito',
    hue='quesito',
    palette='tab20'
)

plt.title('Niterói/2026: DESCq discriminado\n')
plt.xlabel('')
plt.ylabel('')
plt.show()
```

```python
# Gráfico 5: Tuiuti/2017: DESCq discriminado

plt.figure(figsize=(10,6))

sns.barplot(
    data=disc_descq_tui,
    x='descq',
    y='quesito',
    hue='quesito',
    palette='tab20'
)

plt.title('Tuiuti/2017: DESCq discriminado\n')
plt.xlabel('')
plt.ylabel('')
plt.show()
```

As figuras acima são esclarecedoras a respeito da responsabilidade de cada quesito sobre a percepção geral que os jurados tiveram da escola. Fica claro, por exemplo, que o desfile de Niterói em 2026 foi, para o júri, um caso de problemas generalizados por todos os quesitos, sendo o samba-enredo o único que apresenta um tímido DESCq positivo; em outras palavras, na visão dos jurados foi o quesito em que a escola teve menos problemas. ### Comparações temporais possíveis e seus ruídos

Um dos problemas que esperávamos mitigar com a criação do DESCA era a dificuldade da comparação temporal entre escolas que permaneceram por períodos diferentes no Grupo Especial entre 1996 e 2026, marco temporal que analisamos, bem como a complexidade de evolução nas regras que a LIESA estabeleceu e aperfeiçoou ao longo das décadas. A proposta do DESCA, então, era sintetizar a percepção do júri sobre uma escola em um determinado ano, segundo os critérios de avaliação daquele ano; isso *reduz a discrepância conceitual entre as notas que uma escola pode ter tido em dois carnavais diferentes*, tendo em vista a evolução temporal dos critérios de julgamento. Mas ainda resta o maior dos problemas. O Império Serrano, que permaneceu no Grupo Especial apenas 14 dos 30 anos analisados, não pode a princípio ser comparado em uma evolução temporal à Beija-Flor usando apenas a média simples de uma vs. a outra nos 30 anos. Se o objetivo dessa comparação histórica for estabelecer uma medida para o impacto que uma escola causou no júri em relação à outra, precisamos de mais uma nuance: qual foi o DESCA do Império Serrano em cada um dos 14 anos que desfilou no GE e, caso tivesse disputado na mesma categoria em todos os 30 anos da série, como se compararia à Beija-Flor? Matematicamente, isso é apenas o DESCA médio do Império Serrano em 14 anos vs. o DESCA médio da Beija-Flor em 30. Média simples, mas com justiça progressiva: quanto mais anos disputados pela escola, maior é a diluição da sua percepção geral no comportamento dos jurados. Como em qualquer formulação estatística, não nos livramos de todos os problemas: uma escola que só disputou um ano no GE terá provavelmente uma média mais alta que a Beija-Flor em todos os 30 anos, caso neste ano específico seu DESCA seja muito alto. A forma mais elegante de "resolver" isso seria corrigindo o DESCA médio segundo a sua *volatilidade* maior ou menor, medida, talvez, em quantidade de anos disputados no GE; determinaria-se uma quantidade média geral de anos de participação no GE e se utilizaria isso para forçar o DESCA médio de uma escola a tender para o DESCA médio geral; essa técnica é chamada *shrinkage*, e nós não a utilizaremos por enquanto. Primeiro, porque o objetivo dessa exploração não é elaborar um modelo matemático complexo para prever resultados do carnaval (embora, caso no futuro o DESCA possa servir para esse fim, o caminho seja definitivamente esse); segundo, porque de certo modo não se justifica suprimir uma pequena medida de imprevisibilidade inerente à natureza dos desfiles: quero dizer, é perfeitamente possível, mesmo que improvável, que uma escola tenha um péssimo DESCA médio *até hoje* mas, no próximo carnaval, a história seja diferente. Nesse caso, forçar uma tendência ao comportamento médio da escola no sistema, mesmo que apenas para padronizar a tendência *histórica*, pode carregar a análise de um certo determinismo que, a meu ver, não é interessante para o tipo de dados que estamos analisando. Portanto, o único ajuste que faremos para possibilitar a comparação temporal entre as escolas será condicionar o DESCA pela quantidade de anos disputados no GE; para fins de manter a honestidade analítica, deve-se notificar essa distinção na exposição (por exemplo, um gráfico contendo uma comparação temporal entre o Império Serrano e a Beija-Flor entre 1996-2026 deve incluir, necessariamente, pelo menos uma nota de rodapé indicando que a primeira disputou 14 e a segunda 30 anos o Grupo Especial).

Teremos, enfim, a seguinte e definitiva forma de calcular o impacto de uma escola na percepção dos jurados em um dado período:

$$
\bar{DESCA}_{e,p} = \frac{1}{|A_{e,p}|} \sum_{a \in A_{e,p}}{DESCA}_{e,a}
$$

Onde:

- $\bar{DESCA}_{e,p}$ representa o DESCA médio em um dado período;
- $p$ determina o período (por exemplo, $p = [1996,2026]$);
- $e$ representa uma escola específica;
- $A_{e,p}$ é o conjunto dos anos em que a escola disputou o Grupo Especial em um dado período, de forma que ${a \in A_{e,p}}$ são os anos que pertencem a esse conjunto e $|A_{e,p}|$ é o número de elementos no conjunto.

Podemos agora passar às visualizações da comparação entre escolas em séries temporais.
